# Recursion

```python
# factorial number
# 5! = 5*4*3*2*1
# 7! = 7*6*5!

# 1) n*(n-1)  (recursive case)
# 2) 1 or 0  (base case)
```

```python
def factorial(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n*factorial(n-1)
```

```python
factorial(10)
```

Write a recursive Python function fibonacci(n) that returns the nth Fibonacci number.

```python
def fibonacci(n):
  if n <= 1:
    return n
  else:
    return fibonacci(n-1) + fibonacci(n-2)

terms = int(input("Enter the number of terms: "))
if terms <= 0:
  print("Please enter a positive number.")
else:
  print(f"The first {terms} numbers in Fibonacci sequence: ")
  for i in range(terms):
    print(fibonacci(i))
```

```python
# Nested function

def outer(x):
  def inner(y):
    return x+y
  return inner
```

```python
add_five = outer(5)
```

```python
result = add_five(7)
result
```

```python
def mux(x,y):
  return x*y

def square(func, a, b):
  return func(a,b)**2

result = square(mux, 4, 5)
result
```

# Decorators

```python
def myWrapper(func):
  def myInnerFunc():
    print("Inner Wrapper.")
    func()
  return myInnerFunc
```

```python
@myWrapper
def myFunc():
  print("Hello World!")
```

```python
myFunc()
```

```python
def logger(func):
  def wrapper(*args, **kwargs):
    print(f"Calling {func.__name__}")
    return func(*args, **kwargs)
  return wrapper
```

```python
@logger
def greet(name, age):
  print(f"Hello {name}, Age: {age}")
```

```python
greet ("Amrit", age=27)
```
