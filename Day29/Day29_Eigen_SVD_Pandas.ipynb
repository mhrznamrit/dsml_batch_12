```python
import numpy as np
```

```python
A = np.array([
    [1,2],
    [3,4]
])
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigen values: ", eigenvalues)
print("Eigen vector:\n ", eigenvectors)
```

```python
U, S, V = np.linalg.svd(A)
print("U matrix:\n", U)
print("Sigma values:", S)
print("V matrix:\n", V)
```

```python
print(U @ U.T)
```
