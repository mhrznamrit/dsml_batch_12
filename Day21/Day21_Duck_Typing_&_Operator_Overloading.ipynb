```python
# Method  overriding

class PaymentGateway:
    def process_payment(self, amount):
        pass


class CreditCard(PaymentGateway):
    def process_payment(self, amount):
        return f"Processing ${amount} via Credit Card authentication."


class PayPal(PaymentGateway):
    def process_payment(self, amount):
        return f"Processing ${amount} via PayPal secure token."


# Unified interface function
def execute_checkout(payment_method, amount):
    print(payment_method.process_payment(amount))


# Usage
card = CreditCard()
wallet = PayPal()

execute_checkout(card, 250)
execute_checkout(wallet, 250)
```

**Operator Overloading**

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __add__(self, other):
    x = self.x + other.x
    y = self.y + other.y
    return Point(x, y)

  def __sub__(self, other):
    x = self.x - other.x
    y = self.y - other.y
    return Point(x, y)

  def __mul__(self, other):
    x = self.x * other.x
    y = self.y * other.y
    return Point(x, y)

  def __truediv__(self, other):
    x = self.x / other.x
    y = self.y / other.y
    return Point(x, y)

  def __str__(self):
    return f"({self.x}, {self.y})"
```

```python
# Create point object

p1 = Point(7,2)
p2 = Point(9,11)
```

```python
p3 = p1 + p2
print(p3)
```

```python
p4 = p2 - p1
print(p4)
```

```python
p5 = p1 * p2
print(p5)
```

```python
p6 = p2 / p1
print(p6)
```

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __floordiv__(self, other):
    x = self.x // other.x
    y = self.y // other.y
    return Point(x, y)

  def __mod__(self, other):
    x = self.x % other.x
    y = self.y % other.y
    return Point(x, y)

  def __pow__(self, other):
    x = self.x ** other.x
    y = self.y ** other.y
    return Point(x, y)

  def __eq__(self, other):
    x = self.x == other.x
    y = self.y == other.y
    return Point(x, y)

  def __lt__(self, other):
    x = self.x < other.x
    y = self.y < other.y
    return Point(x, y)

  def __gt__(self, other):
    x = self.x > other.x
    y = self.y > other.y
    return Point(x, y)

  def __and__(self, other):
    x = self.x & other.x
    y = self.y & other.y
    return Point(x, y)

  def __or__(self, other):
    x = self.x | other.x
    y = self.y | other.y
    return Point(x, y)

  def __lshift__(self, other):
    x = self.x << other.x
    y = self.y << other.y
    return Point(x, y)

  def __rshift__(self, other):
    x = self.x >> other.x
    y = self.y >> other.y
    return Point(x, y)

  def __str__(self):
    return f"({self.x}, {self.y})"
```

```python
p11 = Point(2, 4)
p12 = Point(3, 4)
```

```python
pp1 = p11 // p12
print(pp1)
pp2 = p11 % p12
print(pp2)
pp3 = p11 ** p12
print(pp3)
pp4 = p11 == p12
print(pp4)
pp5 = p11 < p12
print(pp5)
pp6 = p11 > p12
print(pp6)
pp7 = p11 & p12
print(pp7)
pp8 = p11 | p12
print(pp8)
pp9 = p11 << p12
print(pp9)
pp10 = p11 >> p12
print(pp10)
```

Create a base class Shape with a method area(). Create Rectangle, Circle, and Triangle classes that calculate their areas differently.


```python
import math

class Shape:
  def area(self):
    pass

class Rectangle(Shape):
  def __init__(self, length, breadth):
    self.length = length
    self.breadth = breadth
  def area(self):
    return self.length * self.breadth

class Circle(Shape):
  def __init__(self, radius):
    self.radius = radius
  def area(self):
    return math.pi * (self.radius ** 2)

class Triangle(Shape):
  def __init__(self, base, height):
    self.base = base
    self.height = height
  def area(self):
    return 0.5 * self.base * self.height

def calculate_area(shapes):
  print(f"Area of {shapes.__class__.__name__}: {shapes.area()}")
```

```python
s1 = Rectangle(12, 7)
s2 = Circle(7)
s3 = Triangle(6, 9)

calculate_area(s1)
calculate_area(s2)
calculate_area(s3)
```

Create a base class Employee with a method calculate_salary(). Create FullTimeEmployee and PartTimeEmployee classes that calculate salary using different formulas.

```python
class Employee:
  def calculate_salary(self):
    pass

class FullTimeEmployee(Employee):
  def __init__(self, monthly_salary, bonus):
    self.monthly_salary = monthly_salary
    self.bonus = bonus

  def calculate_salary(self):
    return self.monthly_salary + self.bonus


class PartTimeEmployee(Employee):
  def __init__(self, hourly_rate, hours_worked):
    self.hourly_rate = hourly_rate
    self.hours_worked = hours_worked

  def calculate_salary(self):
    return self.hourly_rate * self.hours_worked

def salary_details(employees):
  emp_type = employees.__class__.__name__
  salary = employees.calculate_salary()
  print(f"{emp_type} Total Salary: Rs.{salary}")
```

```python
ft = FullTimeEmployee(monthly_salary=23000, bonus=9000)
pt = PartTimeEmployee(hourly_rate=50, hours_worked=100)

salary_details(ft)
salary_details(pt)
```

Create a base class Account with a method calculate_interest(amount). Create SavingsAccount, CurrentAccount, and FixedDepositAccount classes with different interest calculations.

```python
class Account:
  def calculate_interest(self, amount):
    pass

class SavingsAccount(Account):
  def calculate_interest(self, amount):
    rate = 0.04
    return amount * rate

class CurrentAccount(Account):
  def calculate_interest(self, amount):
    rate = 0.0
    return amount * rate

class FixedDepositAccount(Account):
  def calculate_interest(self, amount):
    rate = 0.08
    return amount * rate

def display_interest(account, amount):
  acc_type = account.__class__.__name__
  interest = account.calculate_interest(amount)
  print(f"{acc_type} Interest on Rs.{amount}: Rs.{interest}")
```

```python
sav = SavingsAccount()
cur = CurrentAccount()
fix = FixedDepositAccount()

principal = 50000
display_interest(sav, principal)
display_interest(cur, principal)
display_interest(fix, principal)
```
