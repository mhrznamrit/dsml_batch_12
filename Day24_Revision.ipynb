# Recursion

```python
def factorial(n):
    # Base case
    if n == 0 or n == 1:
        return 1
    # Recursive case
    return n * factorial(n - 1)

print(factorial(5))
```

```python
def fibonacci(n):
    # Base cases
    if n == 0:
        return 0
    elif n == 1:
        return 1
    # Recursive case
    return fibonacci(n - 1) + fibonacci(n - 2)

terms = int(input("Enter number of terms: "))
print(f"The first {terms} terms of Fibonacci sequence:")
for i in range(terms):
  print(fibonacci(i), end = " ")
```

```python
def sum_of_digits(n):
    # Base case
    if n < 10:
        return n
    # Recursive case
    return (n % 10) + sum_of_digits(n // 10)

print(sum_of_digits(4562))
```

# Decorators

```python
def decorator(func):
  def wrapper():
    print("Before")
    func()
    print("After")
  return wrapper
```

```python
def hello():
  print("Hello World")
```

```python
func = decorator(hello)
func()
```

Write a decorator login_required that checks whether a user is logged in. If the variable logged_in is True, execute the function; otherwise, print "Access Denied".

```python
logged_in = False

def login_required(func):
  def wrapper():
    if logged_in:
      return func()
    else:
      print("Access Denied")
  return wrapper

@login_required
def dashboard():
    print("Welcome to your dashboard!")
```

```python
print("When user is logged out")
dashboard()
```

```python
print("When user is logged in")
logged_in = True
dashboard()
```

# Iterator and Generator

```python
class Count:
  def __init__(self, limit):
    self.num = 1
    self.limit = limit
  def __iter__(self):
    return self
  def __next__(self):
    if self.num <= self.limit:
      value = self.num
      self.num += 1
      return value
    else:
      raise StopIteration
```

```python
counter = Count(7)
```

```python
for i in counter:
  print(i)
```

```python
def numbers():
  for i in range(1, 6):
    yield i
g = numbers()
print(next(g))
print(next(g))
```

**Student Marks Analyzer**
* Create a dictionary containing student names as keys and their marks as values.
* Find and display:
    * The student with the highest marks
    * The student with the lowest marks
    * The average marks of all students

```python
student_marks ={
    "Anna": 97,
    "Brian": 65,
    "Chris": 22,
    "Dina": 40
}
highest_student = max(student_marks, key=student_marks.get)
lowest_student = min(student_marks, key=student_marks.get)
marks = student_marks.values()
print(f"Student with highest marks: {highest_student} ({student_marks[highest_student]})")
print(f"Student with lowest marks: {lowest_student} ({student_marks[lowest_student]})")
print(f"The average marks is {sum(marks)/len(marks)}")
```

A school has conducted its final examination and stored student names and marks in two separate lists.

students = ["Ram","Sita","Hari","Gita","Shyam","Rita"]

marks = [85, 42, 78, 91, 35, 67]

The school administration wants a small system that can process and analyze the results automatically.

As a Python developer, your task is to develop a program that solves the following requirements using Python's built-in functions such as map(), filter(), reduce(), zip(), and enumerate().

**Questions**
* Display all student names with their serial numbers starting from 1.
* Combine the student names and marks into a single collection containing student records.
* Generate grades for each student (like in SEE)
* Create a list containing only the students who have passed the examination. (marks above 40 is pass)
* Calculate the total marks obtained by all students combined.
* Find the highest mark among all students without using the built-in max() function

```python
from functools import reduce

students = ["Ram","Sita","Hari","Gita","Shyam","Rita"]
marks = [85, 42, 78, 91, 35, 67]

print("--- 1. Serial Numbered Student List ---")
for serial, name in enumerate(students, start=1):
    print(f"{serial}. {name}")

print("\n--- 2. Combined Student Records ---")
records = list(zip(students, marks))
for record in records:
    print(record)

def get_grade(mark):
    if mark >= 90:
        return "A+"
    elif mark >= 80:
        return "A"
    elif mark >= 70:
        return "B+"
    elif mark >= 60:
        return "B"
    elif mark >= 50:
        return "C+"
    elif mark >= 40:
        return "C"
    else:
        return "NG (Not Graded)"

student_grades = list(map(lambda record: (record[0], record[1], get_grade(record[1])), records))
print("\n--- 3. Student Grades ---")
for name, score, grade in student_grades:
    print(f"{name}: Marks = {score}, Grade = {grade}")

print("\n--- 4. Students who Passed ---")
passed_students = list(filter(lambda record: record[1] > 40, records))
for name, score in passed_students:
    print(f"{name}: Marks = {score}")

print("\n--- 5. Total Marks Obtained ---")
total_marks = reduce(lambda acc, score: acc + score, marks)
print(f"Total Marks: {total_marks}")

print("\n--- 6. Highest Mark ---")
highest_mark = reduce(lambda acc, score: acc if acc > score else score, marks)
print(f"Highest Mark: {highest_mark}")
```
