# **Mini Project: Student Management System**

**Project Title:**
Student Management System Using OOP Concepts

**Problem Statement:**
Develop a Student Management System using Python that demonstrates
the concepts of Class, Object, Encapsulation, Inheritance, and Abstraction.
The system should allow users to store student information, calculate
grades, and display student details based on different student categories.

**Requirements**
1. Create an Abstract Class Student

The abstract class should contain:
  * Student ID
  * Student Name
  * A private attribute __marks
  * A method to set marks
  * A method to get marks
  * An abstract method calculate _grade()
  * A method display_details()

```python
from abc import ABC, abstractmethod

class Student(ABC):
  def __init__(self, student_id, name):
    self.student_id = student_id
    self.name = name
    self.__marks = 0.0

  def set_marks(self, marks):
    if 0 <= marks <= 100:
      self.__marks = marks
    else:
      print("Invalid marks! Please enter a value between 0 and 100.")

  def get_marks(self):
    return self.__marks

  @abstractmethod
  def calculate_grade(self):
    pass

  def display_details(self):
    print(f"ID    : {self.student_id}")
    print(f"Name  : {self.name}")
    print(f"Marks : {self.get_marks()}")
    print(f"Grade : {self.calculate_grade()}")
```

```python
class GradeSystem(Student):
  def calculate_grade(self):
    marks = self.get_marks()
    if marks >= 90:
      return "A+"
    elif marks >= 80:
      return "A"
    elif marks >= 70:
      return "B+"
    elif marks >= 60:
      return "B"
    elif marks >= 50:
      return "C+"
    elif marks >= 40:
      return "C"
    elif marks >= 35:
      return "D"
    else:
      return "NG"

class DivisionSystem(Student):
  def calculate_grade(self):
    marks = self.get_marks()
    if marks >= 80:
      return "Distinction"
    elif marks >= 60:
      return "First Division"
    elif marks >= 45:
      return "Second Division"
    elif marks >= 32:
      return "Third Division"
    else:
      return "Fail"
```

```python
student1 = GradeSystem("GS101", "Sunita")
student1.set_marks(69)
student1.display_details()
```

```python
student2 = DivisionSystem("DS901", "Rupa")
student2.set_marks(77)
student2.display_details()
```
