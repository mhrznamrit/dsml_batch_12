```python
# 1. Define bin boundaries and corresponding labels
bins = [0, 12, 19, 59, 100]
labels = ["Child", "Teenager", "Adult", "Senior"]

# 2. Create the new column
titanic_df["AgeGroup"] = pd.cut(
    titanic_df["Age"], bins=bins, labels=labels, right=True
)

# 3. Check value counts with Survived
titanic_df.value_counts(["AgeGroup", "Survived"], ascending=True)
```

```python
titanic_df[titanic_df['AgeGroup'] == "Child"].value_counts("Survived", ascending=True, normalize=True)
```

```python
titanic_df[titanic_df['AgeGroup'] == "Teenager"].value_counts("Survived", ascending=True, normalize=True)
```

```python
titanic_df[titanic_df['AgeGroup'] == "Adult"].value_counts("Survived", ascending=True, normalize=True)
```

```python
titanic_df[titanic_df['AgeGroup'] == "Senior"].value_counts("Survived", ascending=True, normalize=True)
```

```python
titanic_df["FareClass"] = pd.cut(
    titanic_df["Fare"], bins=[0,150,300,450,600], labels=['Low', 'Medium', 'High', 'Very High'], right=True
)
titanic_df.value_counts(["FareClass", "Survived"], ascending=True)
```

```python
titanic_df[titanic_df['FareClass'] == "Low"].value_counts("Survived", ascending=True, normalize=True)
```
