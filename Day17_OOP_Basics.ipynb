# **Class Task**

Create a Python class named Calculator that demonstrates the use of different types of attributes and methods.

Requirements:
1. Create a class attribute named company and assign it any company name.
2. Use the__init__() method to:
    * Ask the user to enter two numbers.
    * Ask the user to enter an operator (+,-,*,/).
3. Create an instance method named calculate() that performs the appropriate operation based on the operator entered by the user.
4. Create a class method named show_company() that displays the company name.
5. Create a static method named welcome() that displays a welcome message.
6. Create an object of the class and call all the required methods.

```python
class Calculator:
  company = "Best"

  def __init__(self):
    self.num1 = int(input("Enter first number: "))
    self.num2 = int(input("Enter second number: "))
    self.op = input("Enter the operation(+,-,*,/):")

  def calculate(self):
    if self.op == '+':
      result = self.num1+self.num2
    elif self.op == '-':
      result = self.num1-self.num2
    elif self.op == '*':
      result = self.num1*self.num2
    elif self.op == '/':
      if self.num2 != 0:
        result = self.num1/self.num2
      else:
        print("Cannot divide by zero.")
    else:
      print("Invalid operator.")
    print("The result is: ", result)

  @classmethod
  def show_company(cls):
    print(f"The company name is {cls.company}")

  @staticmethod
  def welcome():
    print("Warm welcome to you!")
```

```python
ans = Calculator()
ans.calculate()
```

```python
ans.show_company()
```

```python
ans.welcome()
```

# **Student Grade Management System**

Create a Python class named Student.

Requirements:
1. Create a class attribute named school.
2. Use the __init__() method to take:
    * Student name
    * Student marks
3. Create an instance method named check_grade() that displays:
    * A Grade: Marks ≥ 80
    * B Grade: Marks ≥ 60 and < 80
    * C Grade: Marks ≥ 40 and < 60
    * Fail: Marks < 40
4. Create a class method to display the school name.
5. Create a static method to display a welcome message.
6. Create an object and display the student's grade.


```python
class Student:
  school = "Oxford"

  def __init__(self):
    self.name = str(input("Enter the student name: "))
    self.marks = float(input("Enter the student marks: "))

  def check_grade(self):
    if self.marks >= 80:
      grade = "A"
    elif self.marks >= 60:
      grade = "B"
    elif self.marks >= 40:
      grade = "C"
    elif self.marks < 40:
      grade = "Fail"
    else:
      print("Invalid marks!")
    print(f"Student's Name: {self.name} | Grade: {grade}")

  @classmethod
  def show_school(cls):
    print(f"The name of the school is '{cls.school}'")

  @staticmethod
  def welcome_msg():
    print(f"Welcome to School Grade Management System.")
```

```python
Student.show_school()
```

```python
Student.welcome_msg()
```

```python
st1 = Student()
st1.check_grade()
```

# **Bank Account Management System**

Create a Python class named BankAccount.

Requirements:
1. Create a class attribute named bank_name.
2. Use the __init__() method to take:
    * Account holder name
    * Initial balance
3. Create the following instance methods:
    * deposit() → Ask the user for an amount and add it to the balance.
    * withdraw() → Ask the user for an amount and deduct it from the balance if sufficient balance is available.
    * show_balance() → Display the current balance.
4. Create a class method to display the bank name.
5. Create a static method to display banking rules or a welcome message.
6. Create an object and test all methods.

```python
class BankAccount:
  bank_name = "AK Bank"

  def __init__(self):
    self.name = str(input("Enter the account holder name: "))
    self.balance = float(input("Enter the initial balance: "))

  def deposit(self):
    amount = float(input("Enter deposit amount: "))
    self.balance += amount
    print(f"Successfully deposited: Rs.{amount}")
    print(f"New Balance: Rs.{self.balance}")

  def withdraw(self):
    amount = float(input("Enter withdrawal amount: "))
    self.balance -= amount
    print(f"Successfully withdrew: Rs.{amount}")
    print(f"New Balance: Rs.{self.balance}")

  def show_balance(self):
    print(f"Account Holder Name: {self.name} | Current balance: Rs.{self.balance}")

  @classmethod
  def display_name(cls):
    print(f"The bank name is: {cls.bank_name}")

  @staticmethod
  def wel_msg():
    print(f"Welcome to {BankAccount.bank_name}.")
```

```python
BankAccount.display_name()
```

```python
BankAccount.wel_msg()
```

```python
account1 = BankAccount()
account1.deposit()
```

```python
account1.withdraw()
```

```python
account1.show_balance()
```
