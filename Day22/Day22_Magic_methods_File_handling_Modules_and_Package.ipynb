# Magic methods

```python
class BankCustomer:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  # Called by print()
  def __str__(self):
    return f"{self.name} is {self.age} years old."

  # Called by rep
  def __repr__(self):
    return f"{self.name} is {self.age} years old."

  # Called by len
  def __len__(self):
    return self.age

  # Makes object callable like a function
  def __call__(self, gender="Male"):
    return f"{self.name} is {self.age} years old and {gender}."
```

```python
customer = BankCustomer("Alex", 29)
```

```python
# __str__
print(customer)
```

```python
# __repr__
print(repr(customer))
```

```python
# __len__
print(len(customer))
```

```python
# __call__
print(customer())
```

```python
print(customer("Other"))
```

Write a Python program to count the total number of lines, words and characters in a text file.

```python
with open('random.txt', 'r') as file:
  lines = file.readlines()
  line_count = len(lines)
  word_count = sum(len(line.split()) for line in lines)
  char_count = sum(len(line) for line in lines)

  print(f"Lines: {line_count}")
  print(f"Words: {word_count}")
  print(f"Characters: {char_count}")
```

Create a Student class with marks in two subjects. Implement the __add__() magic method so that adding two student objects returns a new student object with combined marks.

```python
class Student:
  def __init__(self, mark1, mark2):
    self.mark1 = mark1
    self.mark2 = mark2
  def __add__(self, other):
    total1 = self.mark1 + other.mark1
    total2 = self.mark2 + other.mark2
    return Student(total1, total2)
  def __str__(self):
    return f"Marks -> Subject 1: {self.mark1}, Subject 2: {self.mark2}"
```

```python
sub1 = Student(95, 87)
sub2 = Student(53, 44)

sub3 = sub1 + sub2
print(sub3)
```

Create a Point class with x and y coordinates. Implement:

  * __str__() to display the point in a readable format.
  * __add__() to add two points together.
  * __repr__() to return a developer-friendly representation of the object.

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __str__(self):
    return f"Point({self.x}, {self.y})"

  def __repr__(self):
    return f"Point(x={self.x}, y={self.y})"

  def __add__(self, other):
    return Point(self.x + other.x, self.y + other.y)
```

```python
p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

print(p3)
print(str(p1))
print(repr(p2))
```
