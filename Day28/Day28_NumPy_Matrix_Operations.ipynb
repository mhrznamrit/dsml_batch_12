```python
# x+y+z=6
# 2x+5y+z=-4
# 2x+3y+8z=5

p = np.array([
    [1,1,1],
    [2,5,1],
    [2,3,8]
])
q = np.array([6,-4,5])
y = np.linalg.solve(p,q)
y
```

```python
x = np.array([
  [1,1],
  [2,1],
  [3,1]
])

y = np.array([2,3,5])

x_t = x.T

mul = x_t @ x

res_inv = np.linalg.inv(mul)

res1 = res_inv @ x_t

final = res1 @ y

final
```

```python
beta = np.linalg.inv(x.T @ x) @ x.T @ y
```

```python
print("Slope (m): ", beta[0])
print("Intercept (c): ", beta[1])
```

# Practice Questions

Create two 3×3 matrices and perform matrix addition.

```python
mat1 = np.array([
    [1,7,0],
    [9,3,8],
    [2,4,5]
])
mat2 = np.array([
    [12,-2,7],
    [0,17,2],
    [-10,6,11]
])
mat_add = mat1 + mat2
mat_add
```

Find the transpose of a 2×4 matrix.

```python
mat3 = np.array([
    [1,5,7,9],
    [2,8,0,3]
])
transpose_mat3 = mat3.T
transpose_mat3
```

Calculate the determinant of a 2×2 matrix.

```python
mat4 = np.array([
    [7,-5],
    [3,9]
])
det = np.linalg.det(mat4)
print("Determinant : ", det)
```

Find the inverse of a matrix and verify the result.

```python
inv_mat4 = np.linalg.inv(mat4)
print("Given matrix:\n", mat4)
print("Inverse matrix:\n", inv_mat4)
verification = np.matmul(mat4, inv_mat4)
print("Verification:\n", np.round(verification))
```

Create a 4×4 identity matrix.

```python
identity_mat = np.eye(4)
print(identity_mat)
```

Find the rank of a given matrix.

```python
mat5 = np.array([
    [1,7,5],
    [20,-3,12],
    [13,0,17],
    [0,23,10]
])
rank = np.linalg.matrix_rank(mat5)
print("Rank of the matrix: ", rank)
```

Compute the trace of a matrix.

```python
mat6 = np.array([
    [1,7,0],
    [9,3,8],
    [2,4,5]
])
trace_val = np.trace(mat6)
print("Given matrix:\n", mat6)
print("Trace of the matrix: ", trace_val)
```

Find eigenvalues and eigenvectors of a 2×2 matrix.

```python
new_mat = np.array([
    [1,5],
    [0,7]
])
eigenvalues, eigenvectors = np.linalg.eig(new_mat)
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

Solve a system of linear equations using np.linalg.solve().
$$\begin{aligned} 2x + y &= 5 \\ x + 3y &= 6 \end{aligned}$$

```python
A = np.array([
    [2,1],
    [1,3]
])
B = np.array([5,6])
x = np.linalg.solve(A,B)
print("x = ", x[0])
print("y = ", x[1])
```

Perform SVD on a 3×3 matrix and print U, S, and VT.

```python
my_matrix = np.array([
    [7,0,-3],
    [1,3,8],
    [-6,5,2]
])
U, S, VT = np.linalg.svd(my_matrix)
print("U Matrix:\n", U)
print("Singular values (S):\n", S)
print("VT (V Transposed) Matrix:\n", VT)
```

Calculate the norm of a vector.

```python
v = np.array([3,4])
norm = np.linalg.norm(v)
print("Vector: ", v)
print("Norm: ", norm)
```

Find the dot product of two vectors.

```python
vect1 = np.array([1,2,3])
vect2 = np.array([4,5,6])
dot_product = np.dot(vect1, vect2)
print("Vector 1: ", vect1)
print("Vector 2: ", vect2)
print("Dot product:", dot_product)
```

Find the cross product of two 3D vectors.

```python
p = np.array([2,3,1])
q = np.array([4,1,2])
cross_product = np.cross(p,q)
print("Vector 1: ", p)
print("Vector 2: ", q)
print("Cross product:", cross_product)
```

Multiply two matrices using @.

```python
matrix1 = np.array([
    [1,2,7],
    [5,4,8]
])
matrix2 = np.array([
    [3,0],
    [8,6],
    [1,9]
])
matrix_mul = matrix1 @ matrix2
print("Matrix 1:\n", matrix1)
print("Matrix 2:\n", matrix2)
print("Matrix multiplication:\n", matrix_mul)
```

Verify that A @ np.linalg.inv(A) produces an identity matrix.

```python
A = np.array([
  [4,7],
  [2,6]
])
A_inv = np.linalg.inv(A)
identity_matrix = A @ A_inv
print("Given matrix:\n", A)
print("Inverse matrix:\n", A_inv)
print("Verification:\n", np.round(identity_matrix))
```
