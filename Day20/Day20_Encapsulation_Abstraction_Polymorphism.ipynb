Create a BankAccount class with:
  * A public account holder name.
  * A protected account type.
  * A private PIN and balance.

Create a SavingAccount class that adds 5% interest to the balance using methods.

```python
class BankAccount:
  def __init__(self, name, pin, balance):
    self.name = name
    self._type = "Saving"
    self.__pin = pin
    self.__balance = balance

  def get_balance(self, pin):
    if pin == self.__pin:
      return self.__balance
    else:
      print("Incorrect PIN")

  def _update_balance(self, amount):
    self.__balance += amount
```

```python
class SavingAccount(BankAccount):
  def add_interest(self):
    balance = self.get_balance(self.pin)
    if isinstance(balance, (int, float)):
      interest = balance * 0.05
      self._update_balance(balance * 0.05)
      print(f"Interest Added: {interest}")
    else:
      print("Not Updated. Balance: ", balance)
```

```python
name = input("Enter your name: ")
balance = float(input("Enter balance: "))
pin = int(input("Set your pin: "))
```

```python
acc = SavingAccount(name, pin, balance)
acc.pin = pin
```

```python
print("Account Details:")
print("Name: ", acc.name)
print("Account Type: ", acc._type)
```

```python
acc.add_interest()
```

```python
entered_pin = int(input("Enter your PIN: "))
print("Balance:",acc.get_balance(entered_pin))
```

Create an abstract BankAccount class with abstract methods deposit() and withdraw(). Implement them in a SavingAccount class.

```python
from abc import ABC, abstractmethod
class BankAccount(ABC):
  def __init__(self, name, balance):
    self.name = name
    self.balance = balance
  @abstractmethod
  def deposit(self, amount):
    pass
  @abstractmethod
  def withdraw(self, amount):
    pass
  def check_balance(self):
    return self.balance

class SavingAccount(BankAccount):
  def deposit(self, amount):
    if amount > 0:
      self.balance += amount
      print(f"Successfully deposited: Rs.{amount}.\nNew Balance: Rs.{self.balance}")
    else:
      print("Deposit amount must be greater than zero.")

  def withdraw(self, amount):
    if amount <= 0:
      print("Withdrawal amount must be greater than zero.")
    elif amount > self.balance:
      print(f"Transaction failed! Insufficient funds. Current balance: Rs.{self.balance}")
    else:
      self.balance -= amount
      print(f"Successfully withdrawn: Rs.{amount}.\nNew Balance: Rs.{self.balance}")
```

```python
account = SavingAccount("am", 12000)
```

```python
account.deposit(5000)
```

```python
account.withdraw(20000)
```

```python
account.withdraw(15000)
```

Write a program to create an ATM class with a private PIN and methods to verify and change the PIN securely.

```python
class ATM:
  def __init__(self, account_name, pin):
    self.name = account_name
    self.__pin = pin
  def verify_pin(self, entered_pin):
    if entered_pin == self.__pin:
      print("PIN verified successfully!")
      return True
    else:
      print("Access Denied: Incorrect PIN!")
      return False
  def change_pin(self, current_pin, new_pin):
    if not self.verify_pin(current_pin):
      print("Failed to change PIN.")
      return
    if len(str(new_pin)) == 4 and isinstance(new_pin, int):
      self.__pin = new_pin
      print("PIN changed successfully!")
    else:
      print("Invaid PIN Format")
```

Write a program to create an abstract Payment class and implement payment processing for different payment methods.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
  def __init__(self, amount):
    self.amount = amount
  @abstractmethod
  def process_payment(self):
    pass

class CreditCard(Payment):
  def __init__(self, amount, card_number):
    super().__init__(amount)
    self.card_number = card_number
  def process_payment(self):
    if self.amount > 0:
      print(f"Payment of Rs.{self.amount} successful.")
      return True
    else:
      print("Payment failed: Invalid amount.")
      return False

class Esewa(Payment):
  def __init__(self, amount, esewa_id):
    super().__init__(amount)
    self.esewa_id = esewa_id
  def process_payment(self):
    if self.amount > 0:
      print(f"Payment of Rs.{self.amount} via {self.esewa_id} successful.")
      return True
    else:
      print("Payment failed: Invalid amount.")
      return False

class Khalti(Payment):
  def __init__(self, amount, khalti_id):
    super().__init__(amount)
    self.khalti_id = khalti_id
  def process_payment(self):
    if self.amount > 0:
      print(f"Payment of Rs.{self.amount} via {self.khalti_id} successful.")
      return True
    else:
      print("Payment failed: Invalid amount.")
      return False
```

```python
card = CreditCard(7500, "727233828328384848")
esewa = Esewa(1200, 9800000000)
khalti = Khalti(2500, 9876543210)
```

```python
card.process_payment()
```

```python
esewa.process_payment()
```

```python
khalti.process_payment()
```

Write a program to create an abstract Animal class and print different sounds using child classes.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def make_sound(self):
    pass

class Dog(Animal):
  def make_sound(self):
    print("Woof! Woof!!")

class Cat(Animal):
  def make_sound(self):
    print("Meow! Meow!!")

class Duck(Animal):
  def make_sound(self):
    print("Quack! Quack!!")
```

```python
dog = Dog()
dog.make_sound()
```

```python
cat = Cat()
cat.make_sound()
```

```python
duck = Duck()
duck.make_sound()
```

Write a program where SavingsAccount and CurrentAccount classes override the interest_rate() method of the Account class.

```python
class Account:
  def __init__(self, account_holder, balance):
    self.account_holder = account_holder
    self.balance = balance
  def interest_rate(self):
    rate = 0.0
    print(f"Standard Account Interest Rate: {rate}")
    return rate

class SavingsAccount(Account):
  def __init__(self, account_holder, balance):
    super(). __init__(account_holder, balance)
  def interest_rate(self):
    rate = 6.5
    print(f"Savings Account Interest Rate for {self.account_holder}: {rate}")

class CurrentAccount(Account):
  def __init__(self, account_holder, balance):
    super().__init__(account_holder, balance)
  def interest_rate(self):
    rate = 1.0
    print(f"Current Account Interest Rate for {self.account_holder}: {rate}")
```

```python
acc = Account("User", 7000)
savings = SavingsAccount("ALex", 25000)
current = CurrentAccount("Business", 75000)
```

```python
acc.interest_rate()
savings.interest_rate()
current.interest_rate()
```

Write a program where Student and Teacher classes override the role() method inherited from the Person class.

```python
from IPython.core.magics import namespace
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def role(self):
    print(f"{self.name} (Age: {self.age}) is a General Person.")

class Student(Person):
  def __init__(self, name, age, student_id):
    super().__init__(name, age)
    self.student_id = student_id
  def role(self):
    print(f"{self.name} (ID: {self.student_id}) is a Student.")

class Teacher(Person):
  def __init__(self, name, age, subject):
    super().__init__(name, age)
    self.subject = subject
  def role(self):
    print(f"{self.name} is a Teacher teaching '{self.subject}'")
```

```python
person = Person("Robin", 45)
student = Student("Penny", 24, "ST4044")
teacher = Teacher("Clint", 33, "Geology")
```

```python
person.role()
student.role()
teacher.role()
```
