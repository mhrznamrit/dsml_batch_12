# List vs Numpy array

```python
#list

import time
start = time.time()
py_list = [i*2 for i in range(100000000)]
print("\n List Operation time: ", time.time()-start)
```

```python
#numpy

start = time.time()
np_array = np.arange(100000000) * 2
print("\nList Operation time: ", time.time()-start)
```

```python
py_list = [1,2,3]
print("Python list multiplication: ", py_list * 3)
```

```python
ny_array = np.array([1,2,3])
print("Numpy array multiplication: ", ny_array * 3)
```

You’ll need to combine array comparisons and logical operators to solve this one. Find out the values in the following array that are greater than 3 AND less than 7.
The output should be a boolean array.

```python
arr = np.array([1,2,5,6,7,9])
result = (arr > 3) & (arr < 7)
print(arr[result])
```
