```python
def count_up(n):
  i = 1
  while i <= n:
    yield i
    i += 1
```

* Write a program that creates a decorator using nested functions and returns the wrapper function.

```python
def myDecorator(func):
  def myWrapper():
    print("Before function execution")
    func()
    print("After function execution")
  return myWrapper

@myDecorator
def greeting():
  print("Hello")

greeting()
```

* Write a program using *args and **kwargs in a decorator so it can work with any function regardless of its parameters.

```python
def universal_decorator(func):
  def wrapper(*args, **kwargs):
    print(f"Calling '{func.__name__}'")
    print(f"Positional args: {args}")
    print(f"Keyword args: {kwargs}")
    return func(*args, **kwargs)
    print(f"Finished executing '{func.__name__}'")
  return wrapper

@universal_decorator
def greet():
  print("Hello, world!")

@universal_decorator
def diff(a, b):
  return a - b

@universal_decorator
def person(name, age, city):
  print(f"{name} is {age} years old and lives in {city}.")
```

```python
greet()
```

```python
result = diff(128,92)
print(f"Result: {result}")
```

```python
person("Hero", age=77, city="Tokyo")
```

* Write a generator function square_generator(n) that yields squares of numbers from 1 to n.

```python
def square_generator(n):
  i = 1
  while i <= n:
    yield i ** 2
    i += 1

gen = square_generator(4)
```

* Develop a mini ATM program that uses custom exceptions, try-except, and finally to handle deposits, withdrawals, and balance checks safely.

```python
class InsufficientFundError(Exception):
  """
  When withdrawal amount exceeds current balance
  """
  pass

def deposit(amount, balance):
  balance += amount
  print(f"Successfully deposited: {amount}")
  print(f"New Balance: {balance}")

def withdraw(amount, balance):
  if amount > balance:
    raise InsufficientFundError(f"Insufficient funds! Available balance: {balance}")
  balance -= amount
  print(f"Successfully withdrawn: {amount}")
  print(f"New Balance: {balance}")

your_balance = 500.00
print("Welcome to MINI ATM")
while True:
  print("\n1. Check Balance \n2. Deposit Money \n3. Withdraw Money \n4. Exit")
  try:
    choice = input("Enter your choice (1-4): ")
    if choice == "1":
      print(f"Currrent Balance: {your_balance}")
    elif choice == "2":
      amt = float(input("Enter deposit amount: "))
      deposit(amt, your_balance)
    elif choice == "3":
      amt = float(input("Enter withdrawal amount: "))
      withdraw(amt, your_balance)
    elif choice == "4":
      print("Thank you for using Mini ATM.")
      break
    else:
      print("Invalid option! Please choose from 1 to 4")
  except ValueError:
    print("Please choose numbers only.")
  except InsufficientFundError as e:
    print(f"Transaction failed: {e}")
  else:
    print("Operation processed smoothly.")
  finally:
    print("Account Updated.")
```
