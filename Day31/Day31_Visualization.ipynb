# Titanic Dataset Practice Questions

Create a box plot of Fare by Passenger Class.

```python
titanic_df.boxplot(
    column="Fare",
    by="Pclass",
    figsize=(10,5)
)
plt.title("Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.suptitle("")
plt.ylabel("Fare")
plt.show()
```

Create a heatmap showing correlations between numerical variables.

```python
import seaborn as sns

plt.figure(figsize=(10, 6))

numeric_df = titanic_df.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numeric_df.corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

Compare the survival count of passengers with and without siblings/spouses (SibSp).

```python
# Using Pandas
titanic_df['Has_SibSp'] = titanic_df['SibSp'].apply(lambda x: 'With SibSp' if x > 0 else 'Without SibSp')

survival_summary = pd.crosstab(titanic_df['Has_SibSp'], titanic_df['Survived'])
print(survival_summary)

survival_summary.plot(kind='bar', figsize=(7, 5), color=['#e74c3c', '#2ecc71'])
plt.title('Survival Count: With vs. Without Siblings/Spouses')
plt.xlabel('Group')
plt.ylabel('Passenger Count')
plt.xticks(rotation=0)
plt.legend(title='Survived', labels=['No (0)', 'Yes (1)'])
plt.show()
```

```python
# Using Seaborn
titanic_df['Has_SibSp'] = titanic_df['SibSp'].apply(lambda x: 'With SibSp' if x > 0 else 'Without SibSp')

plt.figure(figsize=(7, 5))
sns.countplot(data=titanic_df, x='Has_SibSp', hue='Survived', palette='Set2')
plt.title('Survival Count: With vs. Without Siblings/Spouses')
plt.xlabel('Group')
plt.ylabel('Passenger Count')
plt.legend(title='Survived', labels=['No (0)', 'Yes (1)'])
plt.show()
```

Create a violin plot of Age vs Survival.

```python
# Plot the distribution of Age across passenger survival status (Survived)
plt.figure(figsize=(8, 6))

titanic_df['Survived_Label'] = titanic_df['Survived'].map({0: 'Did Not Survive', 1: 'Survived'})

sns.violinplot(
    data=titanic_df,
    x='Survived_Label',
    y='Age',
    palette='Set2',
    inner='quartile'
)

plt.title('Age Distribution by Survival Status')
plt.xlabel('Survival Status')
plt.ylabel('Age')
plt.show()
```

```python
# Split Violin Plot by Gender
plt.figure(figsize=(9, 6))

sns.violinplot(
    data=titanic_df,
    x='Survived_Label',
    y='Age',
    hue='Sex',
    split=True,
    palette='Set1',
    inner='quartile'
)

plt.title('Age vs. Survival Split by Gender')
plt.xlabel('Survival Status')
plt.ylabel('Age')
plt.show()
```

Visualize embarkation port (Embarked) distribution using a count plot.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=titanic_df, x='Embarked', palette='Set2')

plt.title('Passenger Distribution by Embarkation Port')
plt.xlabel('Embarkation Port (S = Southampton, C = Cherbourg, Q = Queenstown)')
plt.ylabel('Passenger Count')
plt.show()
```
